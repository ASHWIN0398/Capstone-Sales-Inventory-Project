```python
import pandas as pd

# Load your dataset 
data = pd.read_csv("../Inventory-Dataset.csv")
data.head()
```

```python
data.info()
```

```python
# Check for missing values in the dataset
missing_values_summary = data.isnull().sum()

# Display columns with missing values
print("Missing values in the dataset:")
print(missing_values_summary[missing_values_summary > 0])
```

```python
# Check and Remove Duplicates
data.drop_duplicates(inplace=True)
```

```python
# Convert 'File_Type' and 'MarketingType' to lowercase for consistency
data['File_Type'] = data['File_Type'].str.lower()
data['MarketingType'] = data['MarketingType'].str.lower()
```

```python
# Fill missing values in 'SoldFlag' and 'SoldCount' columns with 0
data['SoldFlag'].fillna(0, inplace=True)
data['SoldCount'].fillna(0, inplace=True)
print("Missing values in 'SoldFlag' and 'SoldCount' have been filled with 0.")
```

```python
# Rechecking  for missing values in the dataset
missing_values_summary = data.isnull().sum()

# Display columns with missing values
print("Missing values in the dataset:")
print(missing_values_summary[missing_values_summary > 0])
```

```python
#Handling values with 0 in PriceReg and LowUserPrice column

import pandas as pd

# Replace 0 values in 'PriceReg' with the mean
data['PriceReg'] = data['PriceReg'].replace(0, data['PriceReg'].mean())
data['LowUserPrice'] = data['LowUserPrice'].replace(0, data['LowUserPrice'].mean())

#Converting Datatype for ReleaseYear to Int for calculations
data['ReleaseYear'] = data['ReleaseYear'].astype(int)

# Save the cleaned dataset
cleaned_file_path = 'InventoryDataset-Final-Cleaned.csv'
data.to_csv(cleaned_file_path, index=False)

print(f"Cleaned dataset with non-zero 'PriceReg' saved to: {cleaned_file_path}")
```
